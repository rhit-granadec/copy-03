```python
from ultralytics import YOLO
import os
import torch
from contextlib import redirect_stdout, redirect_stderr
import io
```

```python
# GPU and paths
device_id = 5
yaml_file = "data.yaml"
train_path = "./train/images"
val_path   = "./valid/images"

print("Train images:", len(os.listdir(train_path)))
print("Validation images:", len(os.listdir(val_path)))

# Load model
model = YOLO("yolov8n.pt")

f = io.StringIO()
with redirect_stdout(f), redirect_stderr(f):
    results = model.train(
        data=yaml_file,
        epochs=3,
        imgsz=128,
        batch=8,
        device=device_id,
        verbose=False,
        plots=False,
        save=True
    )
```

```python

# Get Metrics
metrics = model.val(data=yaml_file, imgsz=128, batch=8, device=device_id)
print("Validation metrics:", metrics)
print(results.confusion_matrix.summary())
print(metrics.stats)
```
