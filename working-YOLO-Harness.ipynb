```python
import cv2
from ultralytics import YOLO
from ultralytics.utils.plotting import Annotator
import albumentations as A
```

```python
predictOnFrame = 5
```

```python
# Load a model
# model = YOLO("initial-best-4-3-26.pt") # original best model from initial project.
# model = YOLO("first-albumations-run-4-30-26.pt") # abumations attempt at improvement.
model = YOLO("improved-albumations-5-1-2026.pt") # second abumations attempt at improvement
```

```python
# Open the default camera
cam = cv2.VideoCapture(0)

frameCount = predictOnFrame - 1
while True:
    ret, frame = cam.read()

    frameCount = frameCount + 1
    if frameCount >= predictOnFrame :
        frameCount = 0
        [result] = model(source=frame, verbose=False)

    annotator = Annotator(frame, line_width=2)
    for result_num in range(0, result.boxes.shape[0]):
        name = result.names[result.boxes.cls[result_num].item()]+" {conf:.2f}"
        total_label = name.format(conf = (result.boxes.conf[result_num].item()*100))
        bounding_box = result.boxes.xyxy[result_num].tolist()
        annotator.box_label(box=bounding_box, label=total_label)

    # Display the captured frame
    cv2.imshow('Camera', frame)

    # Press 'q' to exit the loop
    if cv2.waitKey(1) == ord('q'):
        break

cv2.destroyAllWindows()

# Release the capture object
cam.release()
```
