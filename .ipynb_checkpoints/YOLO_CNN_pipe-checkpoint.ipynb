```python
from ultralytics import YOLO
import cv2
import os
from YOLO_to_CNN_crop import yolo_to_cnn_crop
from PIL import Image
import torchvision.models as models
import torch
```

```python
def get_free_gpu_safe():
    if not torch.cuda.is_available():
        return torch.device("cpu")
    
    free_mem = []
    for i in range(torch.cuda.device_count()):
        try:
            # torch.cuda.mem_get_info does not require switching the device
            mem_free, _ = torch.cuda.mem_get_info(i)
            free_mem.append((mem_free, i))
        except Exception:
            # GPU completely unusable, skip it
            continue

    if not free_mem:
        return torch.device("cpu")

    _, best_gpu = max(free_mem)
    return torch.device(f"cuda:{best_gpu}")

device = get_free_gpu_safe()
print("Using device:", device)
```

```python
# Load CNN and move it to device
cnn_model = models.resnet18(pretrained=False)
cnn_model.eval()
cnn_model = cnn_model.to(device)  # move model to GPU/CPU

# Load image
img_path = "/work/csse463/202620/03/test/images/000246247_jpg.rf.fb915aef7c063ce2ac971f8de0d8b2c1.jpg"
image = Image.open(img_path).convert("RGB")  

# Run YOLO (YOLO can also use same device)
yolo_model = YOLO('/work/csse463/202620/03/runs/detect/train26/weights/best.pt')
yolo_model.fuse()  # optional for faster inference
yolo_results = yolo_model.predict(img_path, device=device)  # send YOLO to same device

# Process YOLO boxes into CNN-ready tensors
tensors = []
for box in yolo_results[0].boxes.xyxy:  # xyxy = [x1, y1, x2, y2]
    x1, y1, x2, y2 = map(int, box)
    tensor_crop = yolo_to_cnn_crop(image, [x1, y1, x2, y2], input_size=64, pad=0.25)
    tensor_crop = tensor_crop.to(device)  # move each tensor to the same device as CNN
    tensors.append(tensor_crop)

# Stack tensors into a batch
cnn_input = torch.stack(tensors)  # shape: [N, 3, 64, 64]

# Run through CNN
with torch.no_grad():
    outputs = cnn_model(cnn_input)

print("CNN output shape:", outputs.shape)

```
